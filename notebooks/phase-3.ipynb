```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
db_engine = create_engine(
    "postgresql+psycopg2://postgres:airbase234@localhost/test"
)
```

```python
query = "SELECT name, price FROM product WHERE price > 5000"

df = pd.read_sql_query(query, db_engine)

print(df)
```

```python
df = pd.read_sql_table("product", db_engine)

print(df)
```

```python
df.to_sql("product_copy", db_engine, if_exists="replace", index=False)
```

```python
for chunk in pd.read_sql("SELECT * FROM product", db_engine, chunksize=2):
    print(chunk)
```

```python
query = "SELECT * FROM product WHERE price > %(price)s"

df = pd.read_sql(query, db_engine, params={"price": 5000})

print(df)
```

```python
# Read entire table without writing SQL
df_table = pd.read_sql_table("product", db_engine)
print(df_table.head())
```

```python
# Example: price > 5000
query = "SELECT * FROM product WHERE price > %s"
df_filtered = pd.read_sql_query(query, db_engine, params=(5000,))
print(df_filtered)
```

```python
# Example: create a new table with analysis result
df_filtered.to_sql("expensive_products", db_engine, if_exists="replace", index=False)
```

```python
from pandas.io.sql import get_schema

print(get_schema(df_filtered, "expensive_products", con=db_engine))
```
